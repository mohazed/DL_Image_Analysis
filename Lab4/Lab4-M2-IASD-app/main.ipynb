```python
# 1. Mount Google Drive
from google.colab import drive
drive.mount('/content/drive')

import os
import shutil
```

```python
# 2. Define your folder path and local working directory
# Note: Update the DRIVE_FOLDER_PATH if your folder is inside another directory (e.g., /content/drive/MyDrive/Projects/Lab4-M2-IASD-app)
DRIVE_FOLDER_PATH = '/content/drive/MyDrive/Lab4-M2-IASD-app'
LOCAL_WORKING_DIR = '/content/project_root'
```

```python
# 3. Import the folder contents into Colab's local working directory
if os.path.exists(DRIVE_FOLDER_PATH):
    if os.path.exists(LOCAL_WORKING_DIR):
        shutil.rmtree(LOCAL_WORKING_DIR)
    shutil.copytree(DRIVE_FOLDER_PATH, LOCAL_WORKING_DIR)
    print(f"Successfully imported {DRIVE_FOLDER_PATH} to {LOCAL_WORKING_DIR}")
else:
    print(f"Error: Could not find folder at {DRIVE_FOLDER_PATH}. Please check the path.")
```
